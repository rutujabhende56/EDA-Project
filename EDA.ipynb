```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_performance_eda_dataset.csv")
```

```python
print("First 5 Records:")
display(df.head())
```

```python
print("Dataset Shape:", df.shape)
```

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
print("\nDataset Information:")
df.info()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nMissing Value Percentage:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
print("\nStatistical Summary:")
display(df.describe())
```

```python
print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())
```

```python
print("\nUnique Values:")
for column in df.columns:
    print(column, ":", df[column].nunique())
```

```python
print("\nGender Distribution:")
print(df["Gender"].value_counts())
```

```python
numeric_columns = [
    "Age",
    "Study_Hours",
    "Attendance_Percent",
    "Assignments_Completed",
    "Previous_Score",
    "Exam_Score"
]

for column in numeric_columns:
    print(f"\nStatistics for {column}")
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())
```

```python
# Study hours vs Exam score
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Exam_Score"
)
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()


# Attendance vs Exam Score
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Attendance_Percent",
    y="Exam_Score"
)
plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")
plt.show()


# Previous Score vs Exam Score
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Previous_Score",
    y="Exam_Score"
)
plt.title("Previous Score vs Exam Score")
plt.xlabel("Previous Score")
plt.ylabel("Exam Score")
plt.show()

```

```python
correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
display(correlation)

plt.figure(figsize=(9,6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df[numeric_columns])
plt.xticks(rotation=45)
plt.title("Outlier Detection using Boxplot")
plt.show()


# Detect outliers using IQR
for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print("\nColumn:", column)
    print("Number of Outliers:", len(outliers))
```

```python
print("\nAverage Exam Score by Gender:")
print(df.groupby("Gender")["Exam_Score"].mean())


print("\nAverage Exam Score by Assignment Completion:")
print(
    df.groupby("Assignments_Completed")["Exam_Score"]
    .mean()
    .sort_index()
)
```

```python
study_score_correlation = df[
    ["Study_Hours", "Exam_Score"]
].corr().iloc[0,1]

print("\nStudy Hours vs Exam Score Correlation:",
      round(study_score_correlation, 2))

if study_score_correlation > 0:
    print("Result: Positive relationship found.")
else:
    print("Result: No positive relationship found.")
```
